```python
df = pd.read_csv(CHEMIN, sep=";", encoding="UTF-8", keep_default_na=False)
print(df["prenom"].isna().sum())
df[df["prenom"] == "NA"]
```

```python
print(df["periode"].min(), df["periode"].max(), df["periode"].nunique())
print(df["sexe"].value_counts())
# Le but : vérifier que toutes les valeurs sont des multiples de 5, comme l’INSEE l’annonce.
# On cherche des valeurs « louches ».
print(df["valeur"].min(), (df["valeur"] % 5 != 0).sum())
#.mean() fait la moyenne. Vrai compte 1 et Faux compte 0
print((df["valeur"] == 5).mean())
```

```python
totaux = df.groupby(["periode", "sexe"])["valeur"].sum().unstack()
totaux.loc[[1900, 1950, 2000, 2025]]
```

```python
# But. Répondre à : « si un prénom n’a pas de ligne une année, a-t-il été donné 0 fois ? »
# La réponse est non.
g = df.groupby(["sexe", "prenom"])["periode"].agg(["min", "max", "count"])
g["span"] = g["max"] - g["min"] + 1
print((g["span"] != g["count"]).sum(), "sur", len(g))
# le prénoms qui ont des trous 
```

```python
total_sexe_annee = df.groupby(["periode", "sexe"])["valeur"].transform("sum")
df["part_pour_10000"] = df["valeur"] / total_sexe_annee * 10000
df[(df["prenom"] == "MARIE") & (df["sexe"] == 2) & (df["periode"] == 1915)]
```
